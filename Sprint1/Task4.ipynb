```polyglot-notebook
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}; Категория: {category}; Цена: {price}; Количество: {quantity}; Описание: {description}";
    }
}
public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string conta;
    private List<Product> products = new List<Product>();
    public IEnumerable<Product> Products => products;

    public Seller(string name, string employeeId, decimal salary, string conta)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.conta = conta;
    }
    public string Name
    {
        get { return name; }
        set { name = value; }
    }
    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }
    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }
    public string ContactInfo
    {
        get { return conta; }
        set { conta = value; }
    }
    public void AddProduct(Product product)
    {
        products.Add(product);
    }
    public void SellProduct(Product product, int quantity)
    {
        if (products.Contains(product))
        {
            product.UpdateQuantity(-quantity);
        }
        else
        {
            Console.WriteLine("Продавец не имеет этого товара.");
        }
    }
    public string GetSellerInfo()
    {
        return $"Продавец: {name}; ID сотрудника: {employeeId}; Зарплата: {salary}; Контактная информация: {conta}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }

    public void ListProducts()
    {
        Console.WriteLine();
        foreach (Seller seller in sellers)
        {
            Console.WriteLine($"Продавец: {seller.Name}");
            foreach (Product product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
                Console.WriteLine("--------------------");
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}; Адрес: {location}; Часы работы: {storeHours}";
    }
}

Product product1 = new Product("Ноутбук", 999.99m, 10, "Электроника", "Мощный ноутбук для работы и игр.");
Seller seller1 = new Seller("Иван Иванов", "EMP001", 50000.00m, "ivan@example.com");

seller1.AddProduct(product1);
Store store1 = new Store("ТехноМир", "ул. Пушкина, д. 10", "Пн-Сб: 9:00-21:00, Вс: 10:00-18:00");

store1.AddSeller(seller1);

Console.WriteLine("Информация о товаре");
Console.WriteLine(product1.GetProductInfo());
Console.WriteLine($"Общая стоимость: {product1.GetTotalPrice()}");

Console.WriteLine("Информация о продавце");
Console.WriteLine(seller1.GetSellerInfo());
Console.WriteLine();

Console.WriteLine("Информация о магазине");
Console.WriteLine(store1.GetStoreInfo());

Console.WriteLine("Демонстрация продажи");
seller1.SellProduct(product1, 2);
Console.WriteLine($"Количество ноутбуков после продажи: {product1.Quantity}");
```
