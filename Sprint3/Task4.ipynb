```polyglot-notebook
public abstract class Component
{
    public string Name{get;set;}
    public decimal Price{get;set;}
    protected Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
    public virtual string GetInfo()
    {
        return $"Компонент: {Name}, Цена: {Price}";
    }
    public override string ToString()
    {
        return GetInfo();
    }


    public static bool operator ==(Component left, Component right)
    {
        if (ReferenceEquals(left, right)) return true;
        if (left is null || right is null) return false;
        return left.Name == right.Name && left.Price == right.Price;
    }
    
    public static bool operator !=(Component left, Component right)
    {
        return !(left == right);
    }

    
}
public interface IInstalleble
{
    void Install();
}
public class CPU: Component, IInstalleble
{
    public int Cores { get; set; }
    public double Frequency { get; set; }
    public CPU(string name, decimal price, int cores, double frequency) :base(name, price)
    {
        Cores = cores;
        Frequency = frequency;
    }
     public override string GetInfo()
    {
        return $"Процессор: {Name}, Ядер: {Cores}, Частота: {Frequency} GHz, Цена: {Price} руб.";
    }
    public void Install()
    {
        Console.WriteLine($"Установка процессора {Name}");
    }
    public string GetInfo(string manufacturer)
    {
        return $"Процессор: {Name} от {manufacturer}, Ядер: {Cores}, Частота: {Frequency} GHz, Цена: {Price} руб.";
    }
}
public class RAM: Component, IInstalleble
{
    public int Size{get;set;}
    public string Type{get;set;}
    public RAM(string name, decimal price, int size, string type) : base(name, price)
    {
        Size = size;
        Type = type;
    }
    
    public override string GetInfo()
    {
        return $"Оперативная память: {Name}, Объем: {Size} GB, Тип: {Type}, Цена: {Price} руб.";
    }
    public void Install()
    {
        Console.WriteLine($"Установка оперативной памяти {Name}");
    }
    public string GetInfo(bool includeTimings)
    {
        if (includeTimings)
            return $"Оперативная память: {Name}, Объем: {Size} GB, Тип: {Type}, Тайминги: CL16, Цена: {Price} руб.";
        else
            return GetInfo();
    }
}
public class GPU: Component, IInstalleble
{
    public int Memory { get; set; }
    public string MemoryType { get; set; }
    
    public GPU(string name, decimal price, int memory, string memoryType) : base(name, price)
    {
        Memory = memory;
        MemoryType = memoryType;
    }
    
    public override string GetInfo()
    {
        return $"Видеокарта: {Name}, Память: {Memory} GB {MemoryType}, Цена: {Price} руб.";
    }
    
    public void Install()
    {
        Console.WriteLine($"Установка видеокарты {Name}");
    }
    public string GetInfo(bool includeBoostClock)
    {
        if (includeBoostClock)
            return $"Видеокарта: {Name}, Память: {Memory} GB {MemoryType}, Тактовая частота: 1800 MHz, Цена: {Price} руб.";
        else
            return GetInfo();
    }
}
public class HDD: Component, IInstalleble
{
    public int Capacity { get; set; }
    public string Interface { get; set; }
    
    public HDD(string name, decimal price, int capacity, string interfaceType) : base(name, price)
    {
        Capacity = capacity;
        Interface = interfaceType;
    }
    
    public override string GetInfo()
    {
        return $"Жесткий диск: {Name}, Емкость: {Capacity} GB, Интерфейс: {Interface}, Цена: {Price} руб.";
    }
    
    public void Install()
    {
        Console.WriteLine($"Установка жесткого диска {Name}");
    }
    public string GetInfo(bool includeRPM)
    {
        if (includeRPM)
            return $"Жесткий диск: {Name}, Емкость: {Capacity} GB, Интерфейс: {Interface}, Скорость вращения: 7200 RPM, Цена: {Price} руб.";
        else
            return GetInfo();
    }
}
public class Computer
{
    private List<Component> components = new List<Component>();
    
    public void AddComponent<T>(T component) where T : Component, IInstalleble
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Name} успешно добавлен в компьютер");
    }
    
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удален из компьютера");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден в компьютере");
        }
    }
    
    public decimal GetTotalPrice()
    {
        return components.Sum(component => component.Price);
    }
}


List<Component> components = new List<Component>
{
    new CPU("Intel Core i5", 18000, 6, 4.0),
    new RAM("DDR4 16GB", 6000, 16, "DDR4"),
    new GPU("RTX 3060", 35000, 12, "GDDR6"),
    new HDD("SSD 1TB", 8000, 1000, "NVMe")
};


Console.WriteLine();
foreach (var component in components)
{
    Console.WriteLine(component.GetInfo());
}
Console.WriteLine();

Console.WriteLine();

CPU cpu = new CPU("AMD Ryzen 7", 25000, 8, 4.2);
Console.WriteLine("Перегрузка метода GetInfo для CPU:");
Console.WriteLine(cpu.GetInfo());
Console.WriteLine(cpu.GetInfo("AMD"));
Console.WriteLine();

RAM ram = new RAM("DDR5 32GB", 12000, 32, "DDR5");
Console.WriteLine("Перегрузка метода GetInfo для RAM:");
Console.WriteLine(ram.GetInfo()); 
Console.WriteLine(ram.GetInfo(true));
Console.WriteLine();

Component cpu1 = new CPU("Intel Core i5", 18000, 6, 4.0);
Component cpu2 = new CPU("Intel Core i5", 18000, 6, 4.0);
Component cpu3 = new CPU("AMD Ryzen 5", 15000, 6, 3.6);
Component ram1 = new RAM("DDR4 16GB", 6000, 16, "DDR4");

Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 == cpu3: {cpu1 == cpu3}"); 
Console.WriteLine($"cpu1 == ram1: {cpu1 == ram1}"); 
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}"); 
Console.WriteLine();

Console.WriteLine();

Computer myComputer = new Computer();

CPU processor = new CPU("AMD Ryzen 5", 15000, 6, 3.6);
RAM memory = new RAM("DDR4 8GB", 4500, 8, "DDR4");
GPU videoCard = new GPU("GTX 1660", 20000, 6, "GDDR5");

Console.WriteLine("Добавление компонентов:");
myComputer.AddComponent(processor);
myComputer.AddComponent(memory);
myComputer.AddComponent(videoCard);

Console.WriteLine();
foreach (var component in components)
{
    Console.WriteLine(component.ToString());
}

Console.WriteLine($"Общая стоимость компьютера: {myComputer.GetTotalPrice()} руб.");
```
